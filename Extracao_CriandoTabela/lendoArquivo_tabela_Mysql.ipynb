```python
import mysql.connector
import pandas as pd
caminho_vendas_desafio = "C:\\Users\\PC\\Documents\\ComuniDados_____desafios\\vendas_desafio.csv"
```

```python
df_vendas_desafio = pd.read_csv(caminho_vendas_desafio, sep=',')
df_vendas_desafio.head()
```

```python


import mysql.connector
import pandas as pd

conexao = mysql.connector.connect(user='root',password='',host='localhost',database='comuniDados')
cursor = conexao.cursor()

#interage no df e carrega os dados para Banco
cursor.execute('delete from vendas_desafio')

for i, df_vendas_desafio_coluna in df_vendas_desafio.iterrows():

    cursor.execute('''  
        insert into vendas_desafio
        (id_venda,	data_venda,	cliente, produto, categoria, quantidade, preco_unitario, vendedor, cidade, estado) 
        values (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)
    ''',(
        df_vendas_desafio_coluna['id_venda'],
        df_vendas_desafio_coluna['data_venda'],
        df_vendas_desafio_coluna['cliente'],
        df_vendas_desafio_coluna['produto'],
        df_vendas_desafio_coluna['categoria'],
        df_vendas_desafio_coluna['quantidade'],
        df_vendas_desafio_coluna['preco_unitario'],
        df_vendas_desafio_coluna['vendedor'],
        df_vendas_desafio_coluna['cidade'],
        df_vendas_desafio_coluna['estado']
    )
)

conexao.commit()

cursor.close()
conexao.close()

```
